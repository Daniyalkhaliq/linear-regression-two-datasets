```python
import pandas as pd

salary_df = pd.read_csv("datasets/Salary_dataset.csv")

print("Salary dataset loaded successfully!")
display(salary_df.head())

print("Column names:")
print(salary_df.columns.tolist())
```

```python
student_df = pd.read_csv("datasets/Student_performance.csv")

print("Student dataset loaded successfully!")
display(student_df.head())

print("Column names:")
print(student_df.columns.tolist())
```

```python
print("Salary Dataset Columns:")
print(salary_df.columns.tolist())

print("\nStudent Dataset Columns:")
print(student_df.columns.tolist())
```

```python
# Prepare the Salary Dataset
salary_df = salary_df[['YearsExperience', 'Salary']].dropna()

X_salary = salary_df[['YearsExperience']]
y_salary = salary_df['Salary']


# Prepare the Student Dataset
student_df = student_df[['Hours_Studied', 'Exam_Score']].dropna()

X_student = student_df[['Hours_Studied']]
y_student = student_df['Exam_Score']


print("Both datasets are prepared successfully!")

print("Salary dataset shape:", salary_df.shape)
print("Student dataset shape:", student_df.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# Salary Prediction Model
X_train_salary, X_test_salary, y_train_salary, y_test_salary = train_test_split(
    X_salary, y_salary, test_size=0.2, random_state=42
)

salary_model = LinearRegression()
salary_model.fit(X_train_salary, y_train_salary)

salary_predictions = salary_model.predict(X_test_salary)

print("SALARY PREDICTION MODEL")
print("Mean Absolute Error:", mean_absolute_error(y_test_salary, salary_predictions))
print("R2 Score:", r2_score(y_test_salary, salary_predictions))


# Student Exam Score Prediction Model
X_train_student, X_test_student, y_train_student, y_test_student = train_test_split(
    X_student, y_student, test_size=0.2, random_state=42
)

student_model = LinearRegression()
student_model.fit(X_train_student, y_train_student)

student_predictions = student_model.predict(X_test_student)

print("\nSTUDENT EXAM SCORE MODEL")
print("Mean Absolute Error:", mean_absolute_error(y_test_student, student_predictions))
print("R2 Score:", r2_score(y_test_student, student_predictions))
```

```python
import joblib
import os

# Create a folder to store the models
os.makedirs("models", exist_ok=True)

# Save both trained models
joblib.dump(salary_model, "models/salary_model.pkl")
joblib.dump(student_model, "models/student_model.pkl")

print("Both models saved successfully!")
print("Models folder created.")
```
